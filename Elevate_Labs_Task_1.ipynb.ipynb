```python
import pandas as pd

df = pd.read_csv('/content/marketing_campaign.csv', sep='\t')

print("Dataset Shape:", df.shape)

display(df.head())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicates:", df.duplicated().sum())
```

```python

# 1. Import libraries
import pandas as pd

# 2. Store original dataset information
original_rows = len(df)
original_duplicates = df.duplicated().sum()
missing_before = df.isnull().sum().sum()

# 3. Clean column names
df.columns = (
    df.columns.str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

# 4. Remove duplicate records
df = df.drop_duplicates()

# 5. Handle missing Income values using median
df['income'] = df['income'].fillna(df['income'].median())

# 6. Standardize text columns
for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].str.strip()

# 7. Convert customer date into datetime
df['dt_customer'] = pd.to_datetime(
    df['dt_customer'],
    format='%d-%m-%Y',
    errors='coerce'
)

# 8. Check final data quality
print("DATA CLEANING SUMMARY")
print("---------------------")
print("Original Rows:", original_rows)
print("Duplicates Removed:", original_duplicates)
print("Rows After Cleaning:", len(df))
print("Missing Values Before:", missing_before)
print("Missing Values After:", df.isnull().sum().sum())
print("Final Duplicates:", df.duplicated().sum())

print("\nFinal Data Types:")
print(df.dtypes)

# 9. Export cleaned dataset
df.to_csv('cleaned_dataset.csv', index=False)

print("\nCleaned dataset saved successfully!")
```
